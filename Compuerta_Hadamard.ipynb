```python
import math

# 1. Definición del vector de entrada (Estado |0>)
q_input = [
    [1.0],
    [0.0]
]

# 2. Definición de la matriz de transformación (Hadamard H)
val = 1.0 / math.sqrt(2)
H = [
    [val,  val],
    [val, -val]
]

# 3. Algoritmo de multiplicación Matriz (2x2) x Vector (2x1)
def aplicar_compuerta(matriz, vector):
    filas = len(matriz)
    columnas_vec = len(vector[0])

    # Matriz resultante inicializada en cero
    resultado = [[0.0] for _ in range(filas)]

    for i in range(filas):
        for j in range(len(vector)):
            resultado[i][0] += matriz[i][j] * vector[j][0]

    return resultado

# Ejecución

print("Vector de entrada (|0>):")
print(f"|0>: {q_input[0][0]:.4f}")
print(f"|1>: {q_input[1][0]:.4f}")

q_output = aplicar_compuerta(H, q_input)

print("Vector de estado resultante (Superposición):")
print(f"|0>: {q_output[0][0]:.4f}")
print(f"|1>: {q_output[1][0]:.4f}")
```

```python
# 4. Aplicar la compuerta de Hadamard por segunda vez
q_final = aplicar_compuerta(H, q_output)

print("Vector después de la primera Hadamard (Superposición):")
print(f"|0>: {q_output[0][0]:.4f}")
print(f"|1>: {q_output[1][0]:.4f}")

print("\nVector después de la segunda Hadamard (Estado recuperado):")
print(f"|0>: {q_final[0][0]:.4f}")
print(f"|1>: {q_final[1][0]:.4f}")
```
