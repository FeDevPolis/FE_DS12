```python
minutosDisponibles = 5

while minutosDisponibles > 0:
    print(f"Quedan {minutosDisponibles} minutos")
    minutosDisponibles -= 1

print("Tiempo cumplido")
```

| Valor de total| 0 | 1 | 3 | 6 | 10 | 15 | 15 |
| ¿Entra en el tope?| Si | Si | Si | Si | Si | Si | No |

```python
tope = 20
numero = 1
total = 0

while total + numero <= tope:
    total += numero
    print(f"Sumo {numero} -> total: {total}")
    numero += 1

print(f"Total final: {total}")
```

```python
notas = [8, -1, 7, -1, 9]
i = 0
presentes = 0
suma_notas = 0

while i < len(notas):
    nota = notas[i]
    i += 1

    if nota == -1:
        continue

    suma_notas += nota
    presentes += 1

promedio = suma_notas / presentes

print(f"Presentes: {presentes}")
print(f"Promedio: {promedio}")
```

```python
import random
intentos = 0
secreto = random.randint(1,100)

while intentos < 7:
    intentos += 1
    adivina = int(input("Adivina el numero: "))

    if adivina < secreto:
        print("El numero es mas grande")
    elif adivina > secreto:
        print("El numero es mas pequeño")
    elif adivina == secreto:
        print(f"Acertaste en {intentos}")
        break
    else:
        print("Te quedaste sin intentos")
```

```python
totalNota = 0
contador = 0

while True:
    userInput = input("Ingresa tu nota (escribe fin para terminar): ")

    if userInput.lower() == "fin":
        break

    totalNota += int(userInput)
    contador += 1

promedio = totalNota / contador

print("El promedio de las notas es:", promedio)
```
